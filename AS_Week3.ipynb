```python
import cv2
import numpy as np
img = cv2.imread('nightscape.jpg', cv2.IMREAD_GRAYSCALE)
alpha = 1.5
result = alpha * (img.astype(np.float32)-128)+128
result = np.clip(result, 0, 255).astype(np.uint8)
cv2.imwrite('nightscape_contrast15.jpg', result)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

# 원본 이미지
plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original')
plt.axis('off')

# 대비 1.5배 증가 결과
plt.subplot(1, 2, 2)
plt.imshow(result, cmap='gray')
plt.title('Contrast (alpha=1.5)')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. 이미지 읽기
img = cv2.imread('nightscape.jpg', cv2.IMREAD_GRAYSCALE)

# 2. 함수 정의
def adjust_brightness_contrast(img, alpha=1.0, beta=0):
    return cv2.convertScaleAbs(img, alpha=alpha, beta=beta)

# 3. 다양한 값 조합
params = [
    ('원본', 1.0, 0),
    ('밝게', 1.0, 50),
    ('어둡게', 1.0, -50),
    ('대비 증가', 1.5, 0),
    ('대비 감소', 0.5, 0),
    ('밝기+대비', 1.3, 40)
]

results = []
for name, alpha, beta in params:
    out = adjust_brightness_contrast(img, alpha, beta)
    results.append((name, alpha, beta, out))

# 4. 결과 시각화
fig, axes = plt.subplots(2, len(results), figsize=(18, 6))

for i, (name, alpha, beta, out) in enumerate(results):
    axes[0, i].imshow(out, cmap='gray')
    axes[0, i].set_title(f"{name}\n(a={alpha}, c={beta})", fontsize=10)
    axes[0, i].axis('off')

    axes[1, i].hist(out.ravel(), bins=256, range=(0, 255), color='dodgerblue')
    axes[1, i].set_xlim(0, 255)
    axes[1, i].set_ylim(0, 30000)
    axes[1, i].set_xticks([0, 128, 255])
    axes[1, i].set_yticks([0, 15000, 30000])

    if i == 0:
        axes[1, i].set_ylabel('픽셀 수')
        axes[1, i].set_xlabel('픽셀 값')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 이미지 읽기 (그레이스케일)
img = cv2.imread('low_contrast.jpg', cv2.IMREAD_GRAYSCALE)

# 히스토그램 평활화 적용
he_img = cv2.equalizeHist(img)

# 결과 출력
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.title('원본 이미지')
plt.axis('off')
plt.imshow(img, cmap='gray')

plt.subplot(1, 3, 2)
plt.title('평활화 적용 후')
plt.axis('off')
plt.imshow(he_img, cmap='gray')

plt.subplot(1, 3, 3)
plt.title('히스토그램 비교')
plt.hist(img.ravel(), bins=256, range=[0, 256], alpha=0.5, label='원본')
plt.hist(he_img.ravel(), bins=256, range=[0, 256], alpha=0.5, label='평활화 후')
plt.xlim([0, 256])
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 이미지 읽기 (그레이스케일)
img = cv2.imread('low_contrast.jpg', cv2.IMREAD_GRAYSCALE)

# CLAHE 객체 생성
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

# CLAHE 적용
clahe_img = clahe.apply(img)

# 결과 출력
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title('원본 이미지')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.equalizeHist(img), cmap='gray')
plt.title('히스토그램 평활화')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(clahe_img, cmap='gray')
plt.title('CLAHE 적용')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. 컬러 이미지로 읽기
img = cv2.imread('low_contrast.jpg')  # 또는 'nightscape.jpg'

# 2. LAB 색 공간으로 변환 (L: 밝기, A/B: 색상 정보)
img_lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)

# --- (1) 일반 히스토그램 평활화 ---
lab_he = img_lab.copy()
lab_he[:, :, 0] = cv2.equalizeHist(lab_he[:, :, 0])  # L 채널에만 평활화 적용
he_img = cv2.cvtColor(lab_he, cv2.COLOR_LAB2RGB)      # 시각화를 위해 RGB로 변환

# --- (2) CLAHE 적용 ---
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
lab_clahe = img_lab.copy()
lab_clahe[:, :, 0] = clahe.apply(lab_clahe[:, :, 0])  # L 채널에만 CLAHE 적용
clahe_img = cv2.cvtColor(lab_clahe, cv2.COLOR_LAB2RGB)

# 원본도 Matplotlib 출력을 위해 BGR -> RGB 변환
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# 3. 컬러 결과 출력 (cmap='gray' 제거)
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title('원본 이미지')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(he_img)
plt.title('히스토그램 평활화')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(clahe_img)
plt.title('CLAHE 적용')
plt.axis('off')

plt.tight_layout()
plt.show()
```
